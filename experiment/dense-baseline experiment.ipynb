### Extract PDF

```python
from pathlib import Path
import pymupdf
import os
import re
import hashlib
DOCUMENT_PATH = '../data/papers/'
```

```python
def extract_pages(path: Path) -> list[dict]:
    
    with pymupdf.open(path) as doc:
        pages = []
        id = path.stem
        sha256 = hashlib.sha256(path.read_bytes()).hexdigest()
        for i, page in enumerate(doc):
            pages.append({
                'paper_id': id,
                'source_sha256': sha256,
                'page': i + 1,                
                'text': page.get_text()
            })
            
        return pages
```

```python
def chunk_pages(pages: list[dict], words_per_chunks=150) -> list[dict]:
    chunks = []
    for page in pages:
        words = page['text'].split()
        for i in range(0, len(words), words_per_chunks):
            chunks.append({
                'paper_id': page['paper_id'],
                'source_sha256': page['source_sha256'],
                'page': page['page'], 
                'start_word': i,
                'end_word': min(i + words_per_chunks, len(words)),
                'chunk_id' : (
                    f"{page['paper_id']}:{page['source_sha256']}:"
                    f"p{page['page']}:w{i}-{min(i + words_per_chunks, len(words))}"
                    ),
                'text': " ".join(words[i:i+words_per_chunks])
            })
            
    return chunks
```

```python
pages = extract_pages(Path(DOCUMENT_PATH)/'nemotron3super.pdf')
chunks = chunk_pages(pages)

print(f"Chunk count: {len(chunks)}")
print("Example:")
chunks[0]
```

### Embedding

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
EMBEDDING_MODEL = 'BAAI/bge-small-en-v1.5'

model = SentenceTransformer(EMBEDDING_MODEL)
```

```python
token_lengths = [
    len(model.tokenizer(chunk['text'], truncation=False)["input_ids"])
    for chunk in chunks
]

print("Longest chunk:", max(token_lengths))
print("Model limit:", model.max_seq_length)
print("Over limit:", sum(n > model.max_seq_length for n in token_lengths))
```

```python
texts = [chunk['text'] for chunk in chunks]
p_embeddings = model.encode(texts)
print(p_embeddings.shape)
```

### Search

```python
def search(query: str, model, chunks: list, passage_embeddings: list, top_k= 3) -> list:
    if len(query.strip()) < 1:
        raise ValueError("Query must not be blank")
    if top_k < 1:
        raise ValueError("top_k must be at least 1")
    if len(chunks) != len(passage_embeddings):
        raise ValueError("Each chunk must have one passage embedding")
    if len(chunks) == 0:
        return []
    q_embeddings = model.encode(['Represent this sentence for searching relevant passages: '
                                + query])
    cosine_sim = cosine_similarity(q_embeddings, passage_embeddings)
    return sorted([(cosine_sim[0][i], chunks[i]) 
                   for i in range(len(chunks))], 
                   key=lambda x:x[0], 
                   reverse=True)[:top_k]
```

```python
queries = ['What is the model dimension of Nemotron 3 Super?', 'What is Reinforcemnet learning phase of Nemotron 3 Super post-training?']

for q in queries:
    results = search(q,  model, chunks, p_embeddings, 3)
    print(f'Question: {q}')
    for i in range(len(results)):
        print(f"Score: {results[i][0]}")
        print(f"Page: {results[i][1]['page']}")
        print(f"Text: {results[i][1]['text']}\n")
    
```

### Chroma DB

```python
import chromadb

chroma_client = chromadb.PersistentClient('../data/')
collection = chroma_client.get_or_create_collection(name='collection', 
                                metadata={"hnsw:space": "cosine",
                                          "embedding_model": EMBEDDING_MODEL,
                                          "words_per_chunk": 150,
                                          "pipeline_version": "1"})
```

```python
def validate_collection(collection, expected_metadata: dict) -> None:
    metadata = collection.metadata or {}
  
    for field in expected_metadata:
        if metadata.get(field, "<missing>") != expected_metadata[field]:
            raise ValueError(
                            f"{field}: expected {expected_metadata[field]!r}, "
                            f"stored {metadata.get(field, '<missing>')!r}"
                        )
```

```python
validate_collection(collection, 
            expected_metadata={
                "embedding_model": EMBEDDING_MODEL,
                "words_per_chunk": 150,
                "pipeline_version": "1"
            })
```

```python
def upsert_paper(collection, paper_id, chunks, embeddings) -> None:
    validate_collection(collection, expected_metadata={
        "embedding_model": EMBEDDING_MODEL,
        "words_per_chunk": 150,
        "pipeline_version": "1",
    })
    if chunks == []:
        raise ValueError("chunks must have at least 1 element")
    if len(chunks) != len(embeddings):
        raise ValueError("chunks and embeddings must have the same length")
    for chunk in chunks:
        if paper_id != chunk["paper_id"]:
            raise ValueError("Chunk's paper_id does not match the given paper_id")

    old_ids = set(collection.get(where={"paper_id": paper_id})["ids"])
    new_ids = set()
    for i, chunk in enumerate(chunks):
        collection.upsert(
            ids=chunk["chunk_id"],
            embeddings=embeddings[i],
            documents=chunk["text"],
            metadatas={
                "paper_id": chunk["paper_id"],
                "source_sha256": chunk["source_sha256"],
                "page": chunk["page"],
                "start_word": chunk["start_word"],
                "end_word": chunk["end_word"],
            },
        )
        new_ids.add(chunk["chunk_id"])

    stale_ids = old_ids - new_ids
    for chunk_id in stale_ids:
        collection.delete(ids=chunk_id)
```

#### Compare chroma and search function

```python
for i in range(3):
    print(f"Rank: {i+1}")
    print(f"Chroma Score: {1 - chroma_results['distances'][0][i]}" )
    print(f"Search Score: {search_results[i][0]}")
    print(f"Agree: {chroma_results['ids'][0][i] == search_results[i][1]['chunk_id']}")
```

```python
def search_chroma(query: str, model, collection, top_k=3):
    if len(query.strip()) < 1:
            raise ValueError("Query must not be blank")
    if top_k < 1:
        raise ValueError("top_k must be at least 1")
    
    q_embeddings = model.encode(['Represent this sentence for searching relevant passages: '
                                + query])
    hits = collection.query(
            query_embeddings=q_embeddings,
            n_results=top_k)
    return [{"rank": i + 1,
                "chunk_id": hits['ids'][0][i],
                "score": float(1 - hits['distances'][0][i]),
                "text": hits['documents'][0][i],
                "metadata": {
                    "paper_id": hits['metadatas'][0][i]['paper_id'],
                    "source_sha256": hits['metadatas'][0][i]['source_sha256'],
                    "page": hits['metadatas'][0][i]['page'],
                    "start_word": hits['metadatas'][0][i]['start_word'],
                    "end_word": hits['metadatas'][0][i]['end_word']
                }} 
            for i in range(len(hits["ids"][0]))]
        
```

### Run the two-paper benchmark

Run after the two ingestion sections. Results compare the same 224 chunks and retain source checksums and per-query IDs for inspection.


```python
import hashlib
import json
import platform
import sys
import numpy as np

benchmark_chunks = chunks + chunks2
benchmark_embeddings = np.concatenate([p_embeddings, p_embeddings2], axis=0)
benchmark_pages = pages + pages2
page_texts = {
    (page["paper_id"], page["page"]): " ".join(page["text"].split())
    for page in benchmark_pages
}

# A page from another paper must never count as matching evidence.
for case in benchmark:
    assert any(
        case["evidence_anchor"] in page_texts[(case["paper_id"], page_number)]
        for page_number in case["evidence_pages"]
    ), f"Evidence anchor missing for {case['id']}"

expected_ids = {chunk["chunk_id"] for chunk in benchmark_chunks}
assert set(collection.get()["ids"]) == expected_ids, "Collection and benchmark corpus differ"

benchmark_results = []
for case in benchmark:
    hits = search_chroma(case["question"], model, collection, top_k=3)
    baseline_hits = search(
        case["question"], model, benchmark_chunks, benchmark_embeddings, top_k=3
    )
    retrieved = [
        {
            "rank": hit["rank"],
            "chunk_id": hit["chunk_id"],
            "paper_id": hit["metadata"]["paper_id"],
            "page": hit["metadata"]["page"],
            "score": hit["score"],
        }
        for hit in hits
    ]
    page_hit = any(
        hit["paper_id"] == case["paper_id"]
        and hit["page"] in case["evidence_pages"]
        for hit in retrieved
    )
    same_ranking = (
        [hit["chunk_id"] for hit in hits]
        == [chunk["chunk_id"] for _, chunk in baseline_hits]
    )
    benchmark_results.append({
        "id": case["id"],
        "question": case["question"],
        "reference_answer": case["reference_answer"],
        "paper_id": case["paper_id"],
        "evidence_pages": case["evidence_pages"],
        "retrieved": retrieved,
        "paper_page_hit_at_3": page_hit,
        "same_chunk_ranking": same_ranking,
    })
    locations = [(hit["paper_id"], hit["page"]) for hit in retrieved]
    print(f"{case['id']} | retrieved {locations} | hit {page_hit} | same ranking {same_ranking}")

corpus_versions = {
    paper_id: hashlib.sha256(
        (Path(DOCUMENT_PATH) / f"{paper_id}.pdf").read_bytes()
    ).hexdigest()
    for paper_id in sorted({chunk["paper_id"] for chunk in benchmark_chunks})
}
assert all(
    chunk["source_sha256"] == corpus_versions[chunk["paper_id"]]
    for chunk in benchmark_chunks
), "Source PDF changed since extraction"

hit_count = sum(result["paper_page_hit_at_3"] for result in benchmark_results)
benchmark_report = {
    "benchmark_version": "two-paper-dev-v2",
    "corpus_versions": corpus_versions,
    "benchmark_sha256": hashlib.sha256(
        json.dumps(benchmark, sort_keys=True).encode()
    ).hexdigest(),
    "model": EMBEDDING_MODEL,
    "query_prefix": "Represent this sentence for searching relevant passages: ",
    "words_per_chunk": 150,
    "chunk_count": len(benchmark_chunks),
    "top_k": 3,
    "python": sys.version.split()[0],
    "platform": platform.platform(),
    "paper_page_hit_at_3": hit_count / len(benchmark_results),
    "results": benchmark_results,
}
print(f"Paper-page Hit@3: {hit_count}/{len(benchmark_results)}")
print("Matching rankings:", sum(r["same_chunk_ranking"] for r in benchmark_results))
```
